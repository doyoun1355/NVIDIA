```python
!uv pip install ultralytics
import ultralytics
ultralytics.checks()

!yolo predict model=yolo11n.pt source='https://ultralytics.com/images/zidane.jpg'

# ▶️ 1) 유튜브 영상 다운로드
!pip install yt_dlp --quiet
import yt_dlp
url = 'https://youtube.com/shorts/P22fW-GBFI4?si=CQCVuOungH1eutmJ'
ydl_opts = {'outtmpl': 'input_video.mp4', 'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/mp4'}
with yt_dlp.YoutubeDL(ydl_opts) as ydl: ydl.download([url])

from ultralytics import YOLO
import cv2

# YOLOv11 모델 로드 (사전에 업로드한 yolo11n.pt 사용)
model = YOLO('yolo11n.pt')

# 영상 열기
cap = cv2.VideoCapture("input_video.mp4")
width, height = int(cap.get(3)), int(cap.get(4))
fps = cap.get(cv2.CAP_PROP_FPS)

# 결과 저장용 비디오 설정
out = cv2.VideoWriter('output_vehicle.mp4', cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame_count += 1
    print(f'🚘 차량 탐지 중... 프레임 {frame_count}')

    # YOLO 차량 탐지
    results = model(frame, conf=0.4)

    # 시각화 결과 추출
    annotated_frame = results[0].plot()

    # 프레임 저장
    out.write(annotated_frame)

cap.release()
out.release()
cv2.destroyAllWindows()

print("🎉 차량 인식 완료! 저장된 영상: output_vehicle.mp4")

from google.colab import files
files.download('output_vehicle.mp4')
```
